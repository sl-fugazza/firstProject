```python
dataframe.describe()

# there are some issues: 
# 1) glucose cannot be zero (the patient should be dead) -> Let's treat these values as NaN
# 2) Same thing for bloodPressure, skinThickness, BMI
# 3) Fasting Insulin level should be > 2 usually ... I guess also Insulin 0 values must be corrected
```

```python
# dataframe["Insulin"].hist(bins=100)      # This confirms that Insulin = 0 is a missing value 
# dataframe["Glucose"].hist(bins=30)       # same thing 
# dataframe["BloodPressure"].hist(bins=30) # same thing 
# dataframe["SkinThickness"].hist(bins=30) # same thing  
# dataframe["BMI"].hist(bins=30)           # same thing  

cols = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]

dataframe[cols] = dataframe[cols].astype(float)

dataframe[cols] = dataframe[cols].replace(0.0, np.nan)

# let's give to them a median value -> we're not forced to remove us and lost entries

imputer = skl.impute.SimpleImputer(strategy='median')
dataframe[cols] = imputer.fit_transform(dataframe[cols])

dataframe

# dataframe["BMI"].hist(bins=30) # check
```

```python
corr = dataframe.corr()

plt.figure(figsize=(12,10))
sbn.heatmap(corr,  vmin=-1, vmax = 1, center = 0,
            cmap = sbn.diverging_palette(20,220, n=200),
            annot=True, square = True)
plt.show()

corr
```

```python
# checks multicollinearity 

from statsmodels.stats.outliers_influence import variance_inflation_factor 

def calc_vif(X): # X: the data frame
    vif = pd.DataFrame()
    vif["variables"] = X.columns
    vif["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])] 
    vif.sort_values(by=['VIF'], ascending = True)
    return(vif)
    
calc_vif(dataframe.drop(columns=['Outcome']))  # no target variable
```

```python
# Data Visualisation: old | middle-aged | young people VS Outcome

young_df = dataframe[dataframe["Age"]<30]
middle_df = dataframe[(dataframe["Age"] >= 30) & (dataframe["Age"] < 60)]
old_df = dataframe[dataframe["Age"]>=60]

n_young = young_df.shape[0]
n_middle = middle_df.shape[0]
n_old = old_df.shape[0]

# young_df.groupby("Outcome").count()['BMI'].to_frame("Number young").plot(kind='bar')
# middle_df.groupby("Outcome").count()['BMI'].to_frame("Number middle").plot(kind='bar')
# old_df.groupby("Outcome").count()['BMI'].to_frame("Number old").plot(kind='bar')

young_df = young_df.groupby("Outcome").count()["BMI"].to_frame("n_young")
middle_df = middle_df.groupby("Outcome").count()['BMI'].to_frame("n_middle")
old_df = old_df.groupby("Outcome").count()['BMI'].to_frame("n_old")

age_df = pd.DataFrame([])
age_df["young"] = young_df["n_young"]
age_df["middle"] = middle_df["n_middle"]
age_df["old"] = old_df["n_old"]
age_df["tot"] = age_df["young"] + age_df["middle"] + age_df["old"] 

n_set = age_df.sum()["tot"]

age_df["%-y"] = 100*age_df["young"]/age_df["tot"]
age_df["%-m"] = 100*age_df["middle"]/age_df["tot"]
age_df["%-o"] = 100*age_df["old"]/age_df["tot"]

age_df[["%-y", "%-m", "%-o"]].plot(kind='bar')

age_df
```

```python
# feature scaling 

# robust scaler for all features -> neither gaussian nor uniform distributions

rob_scaler = skl.preprocessing.RobustScaler()

allCols = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 
                'BMI', 'DiabetesPedigreeFunction', 'Age']

df[allCols] = rob_scaler.fit_transform( dataframe[allCols] )
df["Outcome"] = dataframe["Outcome"]
df
```

```python
# ML - Classification

X = df.drop(columns=['Outcome'])
y = df['Outcome']
 
X_train, X_test, y_train, y_test = skl.model_selection.train_test_split(X, y, 
                                                test_size=0.2, random_state=0, stratify=y)
# stratifiy=Y -> equal 0 and 1 values in both train and test subset
# it maintein the correct proportion between train and test set

model = skl.linear_model.LogisticRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

```

```python
print("Confusion Matrix:")
print(skl.metrics.confusion_matrix(y_test, y_pred))
print("Classification Report:")
print(skl.metrics.classification_report(y_test, y_pred))
```

```python
# feature importance 

model = skl.linear_model.LogisticRegression()
fitted_model = model.fit(X_train, y_train)

permutation_score = skl.inspection.permutation_importance(fitted_model,
                                                          X_test, y_test, scoring='r2', n_repeats=100)

importance_df = pd.DataFrame({'Feature': X.columns, 'Score_Decrease':permutation_score.importances_mean})
importance_df
```
